Task : Outlier Decision

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

age_outliers = df[(df['Age'] < lower) | (df['Age'] > upper)]

print(age_outliers[['Age']])
```

```python
Q1 = df['Unit_Price'].quantile(0.25)
Q3 = df['Unit_Price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

price_outliers = df[(df['Unit_Price'] < lower) | (df['Unit_Price'] > upper)]

print(price_outliers[['Unit_Price']])
```

```python
Q1 = df['Total_Amount'].quantile(0.25)
Q3 = df['Total_Amount'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

amount_outliers = df[(df['Total_Amount'] < lower) | (df['Total_Amount'] > upper)]

print(amount_outliers[['Total_Amount']])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv("cleaned_ecommerce_dataset.csv")

# Light pink color
pink = "#FFB6C1"


# 1. Histogram of Age
plt.figure(figsize=(7, 4))
sns.histplot(df['Age'], bins=20, kde=True, color=pink)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# 2. Histogram of Total_Amount
plt.figure(figsize=(7, 4))
sns.histplot(df['Total_Amount'], bins=30, kde=True, color=pink)

plt.title("Distribution of Total Amount")
plt.xlabel("Total Amount")
plt.ylabel("Frequency")
plt.show()


# 3. KDE Plot of Unit_Price
plt.figure(figsize=(7, 4))
sns.kdeplot(df['Unit_Price'], fill=True, color=pink)

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Density")
plt.show()


# 4. Boxplot of Total_Amount
plt.figure(figsize=(7, 4))
sns.boxplot(x=df['Total_Amount'], color=pink)

plt.title("Boxplot of Total Amount")
plt.xlabel("Total Amount")
plt.show()


# 5. Countplot of Product_Category
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Product_Category'], color=pink)

plt.title("Product Category Distribution")
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()


# 6. Countplot of Payment_Method
plt.figure(figsize=(8, 5))
sns.countplot(x=df['Payment_Method'], color=pink)

plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()


# 7. Countplot of Gender
plt.figure(figsize=(6, 4))
sns.countplot(x=df['Gender'], color=pink)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

   The chart compares product categories across genders. The number of customers from each gender varies across different product categories.

   The chart compares return status across product categories. The number of returned and non-returned orders differs between product categories.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv("cleaned_ecommerce_dataset.csv")

# Select numerical columns
num_cols = [
    'Age',
    'Unit_Price',
    'Quantity',
    'Discount',
    'Rating',
    'Delivery_Days',
    'Total_Amount'
]

# Create correlation matrix
correlation = df[num_cols].corr()

# Display correlation matrix
print("Correlation Matrix:")
print(correlation)

# Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Matrix")
plt.show()


# Remove self-correlations
corr_pairs = correlation.where(
    ~np.eye(correlation.shape[0], dtype=bool)
)


# 1. Strongest positive correlation
positive = corr_pairs.stack().sort_values(ascending=False)

print("\n1. Strongest Positive Correlation:")
print(positive.head(1))


# 2. Strongest negative correlation
negative = corr_pairs.stack().sort_values(ascending=True)

print("\n2. Strongest Negative Correlation:")
print(negative.head(1))


# 3. Weak correlations
weak = corr_pairs.abs().stack().sort_values()

print("\n3. Weak Correlations:")
print(weak.head(5))


# 4. Variables useful for predicting Total_Amount
total_corr = correlation['Total_Amount'].drop('Total_Amount')

total_corr = total_corr.sort_values(
    key=abs,
    ascending=False
)

print("\n4. Variables useful for predicting Total_Amount:")
print(total_corr)
```


Which features appear to have the strongest relationship with Total Amount?


The features that appear to have the strongest relationship with Total_Amount are Unit_Price and Quantity.

Unit_Price has a strong positive relationship with Total_Amount because a higher product price generally increases the total order amount.

Quantity also has a positive relationship with Total_Amount because purchasing more units increases the total amount.

Discount may have a weaker relationship with Total_Amount, while Age, Rating, and Delivery_Days appear to have relatively weaker relationships.

Therefore, Unit_Price and Quantity are the most useful features for predicting Total_Amount.
